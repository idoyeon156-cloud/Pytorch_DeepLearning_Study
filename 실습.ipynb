```python
import torch

#1차원 텐서 선언과 프린트, 슬라이싱
t=torch.tensor([1., 3., 5., 7., 9., 4])
print(t)

print(t.dim())
print(t.shape)
print(t.size())

print(t[0], t[3], t[-1])
print(t[1:4], t[3:-1])
print(t[:3], t[4:])
```
